```python
from training_loop import *

def try_hyperparameters(batch_size: int = 48, total_tokens_processed: int = 327_680_000):
    lr_scale_by_batch_size = batch_size / 48
    adamw_params = AdamWParams(1e-3 * lr_scale_by_batch_size, 0.01, (0.9, 0.999), 1e-8, 1e-4 * lr_scale_by_batch_size, 0.02, 1.0) # Best params from experiments_lr.ipynb
    tiny_stories_params = TransformerParams(
        vocab_size = 10000,
        context_length = 256,
        num_layers = 4,
        d_model = 512,
        num_heads = 16,
        d_ff = 1344,
        theta = 10000
    )
    num_iterations = total_tokens_processed // (batch_size * tiny_stories_params.context_length)
    print(f"Total iterations: {num_iterations}")
    train_model(
        train_data_path='../data/TinyStoriesV2-GPT4-train.dat',
        val_data_path='../data/TinyStoriesV2-GPT4-valid.dat',
        params=tiny_stories_params,
        adamw_params=adamw_params,
        batch_size=batch_size,
        num_iterations=num_iterations,
        log_interval=25,
        checkpoint_interval=num_iterations+1,
        compile = True,
        checkpoint_root=None,
        dtype=torch.float32,
        device=torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    )
```
